```python
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

from matplotlib.ticker import FuncFormatter
```

```python
# Load the local copy using 'Name' as the index
df = pd.read_csv("football_players.csv", 
                 index_col="Name",
                low_memory=False)

# Display the first 5 rows
df.head()
```

```python
print(df.info())
```

```python
# Numeric attributes clean up for visualization
numeric_cols = [
    'Age', 'Overall', 'Acceleration', 'Ball control', 'Dribbling', 
    'Finishing', 'Short passing', 'Stamina', 'Strength', 'Vision', 'Shot power'
]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')

df['Primary_Position'] = df['Preferred Positions'].str.strip().str.split().str[0]
```

```python
class FootballVisualizer:
    """
    A reusable visualization class for football player analysis.

    Features
    --------
    • Publication-quality plots
    • White background
    • Light grid
    • Removed top/right spines
    • Automatic styling
    • Histogram
    • Bar Plot
    • Scatter Plot
    • Correlation Heatmap
    """

    def __init__(self, df, primary_color='#1f77b4', grid_color='0.75', figsize=(8,4)):
        self.df = df
        self.primary_color = primary_color
        self.grid_color = grid_color
        self.figsize = figsize
        
        sns.set_theme(style="white")
        plt.rcParams.update({

            "figure.figsize": self.figsize,
            "axes.titlesize": 16,
            "axes.titleweight": "bold",
            "axes.labelsize": 12,
            "xtick.labelsize": 10,
            "ytick.labelsize": 10,

            "legend.fontsize": 10

        })
        
    ####################################################################
    # Private styling method
    ####################################################################

    def _style_axes(self,ax,title,xlabel,ylabel):
        ax.set_title(title, pad=12)
        ax.set_xlabel(xlabel)
        ax.set_ylabel(ylabel)

        ax.set_axisbelow(True)
        ax.grid(True, color=self.grid_color, linestyle="--", linewidth=0.8, alpha=0.4)

        for spine in ax.spines.values():
            spine.set_visible(False)
        ax.tick_params(length=0, axis='both')
    
        plt.tight_layout()

    ####################################################################
    # Histogram
    ####################################################################

    def histogram(
        self,
        column,
        bins=25,
        color="steelblue",
        kde=True):

        fig, ax = plt.subplots()

        sns.histplot(
            data=self.df,
            x=column,
            bins=bins,
            kde=kde,
            color=color,
            edgecolor="black",
            alpha=0.7,
            ax=ax
        )

        self._style_axes(
            ax,
            f"Distribution of {column}",
            column,
            "Frequency"
        )

        plt.show()

    ####################################################################
    # Bar Plot
    ####################################################################

    def barplot(
        self,
        x,
        y,
        order=None,
        color="steelblue",
        annotate=False
    ):

        fig, ax = plt.subplots()

        sns.barplot(

            data=self.df,

            x=x,

            y=y,

            order=order,

            color=color,

            errorbar=None,

            ax=ax

        )

        if annotate:

            for container in ax.containers:

                ax.bar_label(
                    container,
                    fmt="%.1f",
                    fontsize=10
                )

        self._style_axes(

            ax,

            f"{y} by {x}",

            x,

            y

        )

        plt.xticks(rotation=30)

        plt.show()

    ####################################################################
    # Scatter Plot
    ####################################################################

    def scatterplot(
        self,
        x,
        y,
        hue=None,
        size=None,
        alpha=0.7
    ):

        fig, ax = plt.subplots()

        sns.scatterplot(

            data=self.df,

            x=x,

            y=y,

            hue=hue,
            palette = 'viridid' if hue else None,
            size=size,
            alpha=alpha,
            edgecolor="black",
            linewidth=0.4,
            ax=ax
        )

        self._style_axes(ax, f"{y} vs {x}", x, y)

        plt.show()

    ####################################################################
    # Heatmap
    ####################################################################

    def heatmap(
        self,
        columns=None,
        cmap="YlGnBu",
        annot=True
    ):

        if columns is None:

            corr = self.df.select_dtypes(
                include=np.number
            ).corr()

        else:

            corr = self.df[columns].corr()

        fig, ax = plt.subplots(
            figsize=(12, 9)
        )

        sns.heatmap(
            corr,
            cmap=cmap,
            annot=annot,
            fmt=".2f",
            linewidths=.5,
            linecolor='white',
            cbar = True,
            square=True,

            cbar_kws={
                "shrink": .8
            },
            ax=ax

        )

        ax.set_title(
            "Correlation Heatmap",
            fontsize=20,
            pad=12,
            weight="bold"
        )

        plt.tight_layout()

        plt.show()
```

```python
viz = FootballVisualizer(df)
```

__Age Distribution__

```python
viz.histogram("Age")
```

__Bar Plot__

```python
position_rating = (
    df.groupby("Primary_Position")["Overall"]
      .mean()
      .sort_values(ascending=False)
      .head(10)
      .reset_index()
)
```

```python
viz = FootballVisualizer(position_rating)

viz.barplot(
    x="Primary_Position",
    y="Overall",
    annotate=True
)
```

__Heatmap__

```python
viz = FootballVisualizer(df)

viz.scatterplot(
    x="Overall",
    y="Stamina",
    hue="Preferred Foot"
)
```
