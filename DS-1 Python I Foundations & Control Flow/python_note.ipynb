```python
def get_low_stock_alerts(inventory_list):
    """
    Returns a low stock alert based on inventory levels.

    Args:
        inventory_list (list): A list of dictionaries containing item names and stock levels.

    Returns:
        str: A stock alert message.
    """

    low_stock_items = [
        item["name"]
        for item in inventory_list
        if item.get("stock", 0) < 10
    ]
    
    if not low_stock_items:
        return "All stock levels are good."

    return f"Low stock warning for: {', '.join(low_stock_items)}"
```

```python
inventory = [
    {"name": "Laptop", "stock": 15},
    {"name": "Keyboard", "stock": 5},
    {"name": "Mouse", "stock": 3},
    {"name": "Monitor", "stock": 12}
]

print(get_low_stock_alerts(inventory))
```

```python
import string
import re

def count_word_frequency(text):
    # Clean the text: convert to lowercase and remove punctuation
    text = text.lower().translate(
        str.maketrans("", "", string.punctuation)
    )

    # Remove numbers (if any)
    text = re.sub(r"\d+", '', text)
    
    # Split the text into individual words
    words = text.split()
    
    # Build the frequency dictionary
    # If word is in dictionary, increment count; otherwise set to 1
    word_count = {}
    for word in words:
        word_count[word] = word_count.get(word, 0) + 1

    return word_count
```
