```python
text = """
    Human history is often framed through the lens of physical monuments—the towering pyramids of Giza, the vast networks of Roman aqueducts, or the sprawling steel skylines of modern metropolises. Yet, running parallel to these physical marvels is an invisible, silent architecture: the realm of mathematical thought. Unlike brick and mortar, which inevitably weather, fracture, and erode under the weight of time, a mathematical truth remains unaltered across millennia. The geometric principles mapped out on sandboards in ancient Greece or calculated on clay tablets in Babylon hold the exact same logical validity today as they will a thousand years in the future.

At its core, abstract mathematics is not merely a tool for calculation or ledger-keeping; it is a specialized language designed to uncover hidden structure in a seemingly chaotic universe. When we observe the world around us, nature presents a dizzying array of sensory inputs—the turbulent swirl of a river around a boulder, the unpredictable fluctuations of weather patterns, or the delicate arrangement of seeds inside a sunflower head. To the uninitiated eye, these events appear disparate and disconnected. However, when viewed through the analytical lens of patterns, symmetry, and quantitative relationships, underlying order begins to emerge.

Consider the phenomenon of growth. The spiral of a nautilus shell, the arrangement of pinecone scales, and the branching of galaxy arms are not random occurrences. They are physical manifestations of recursive numerical sequences and optimal geometric packing. By translating these physical forms into formal equations, mathematicians can strip away the noise of the physical material—the calcium carbonate of the shell or the wood fiber of the pinecone—to isolate the pure, underlying dynamic rule. This process of abstraction allows us to model non-physical systems using the exact same underlying logic, whether predicting financial market volatility, optimizing network traffic across global data channels, or training complex algorithmic models.

What makes this intellectual pursuit so profound is its unexpected utility. History is filled with instances where pure, theoretical concepts—developed entirely out of intellectual curiosity without any immediate practical application—later became the indispensable foundation for technological revolutions. When 19th-century mathematicians began exploring non-Euclidean geometries, bending the rules of flat space to imagine curved manifolds, it was viewed by many as an elegant mental exercise. Decades later, these exact spatial concepts provided the framework needed to describe the curvature of spacetime in modern physics.

To engage with abstract systems is to participate in an ongoing, global dialogue that transcends cultural, temporal, and linguistic boundaries. It requires a unique blend of creative imagination and rigorous logical discipline. One must hold hypothetical worlds inside their mind, test their internal boundaries through proof, and accept with humility wherever the logic leads. In an era increasingly driven by rapid information streams and automated systems, cultivating the capacity for deep, structured, and deliberate abstract reasoning remains one of the most powerful endeavors of the human mind.

"""
```

#### __Text Preprocessing__

__Preprocessing converts raw text into a format suitable for analysis__

__Convert to lowercase__

```python
text = text.lower()
```

__Remove Punctuation__

```python
text = text.translate(str.maketrans("", "", string.punctuation))
```

__Remove numbers (if any)__

```python
text = re.sub(r"\d+", '', text)
```

#### __Tokenization__

__Split text into individual words__

```python
words = word_tokenize(text)
words[:10]
```

__Sentences Tokenization__

```python
sentences = sent_tokenize(text)
sentences
```

#### Remove stopwords

```python
stop_words = set(stopwords.words('english'))
filtered_words = [word for word in words if word not in stop_words]
```

#### __Word Freqeuncy__

```python
word_freq = Counter(filtered_words)

print(word_freq.most_common(10))
```

#### __Lemmatization__

Reduce words to their dictionary form.

```python
nlp = spacy.load("en_core_web_sm")

doc = nlp(text)

lemmas = [token.lemma_ for token in doc]
print(lemmas)
```

#### __Named Entity Recognition (NER)__

```python
for ent in doc.ents:
    print(ent.text, ent.label_)
```

#### __Sentiment Analysis__

Using TextBlob

```python
blob = TextBlob(text)

blob.sentiment
```

```python
bigrams = list(ngrams(filtered_words,2))
bigrams[:10]
```

#### __TF-IDF__

Find important words.

```python
vectorizer = TfidfVectorizer(stop_words="english")

X = vectorizer.fit_transform([text])
```

```python
features = vectorizer.get_feature_names_out()

print(features)
```

```python
import pandas as pd

df = pd.DataFrame({
    "Word": vectorizer.get_feature_names_out(),
    "TF-IDF Score": X.toarray()[0]
})

df
```

```python
df = df.sort_values(by="TF-IDF Score", ascending=False)

df.head(10)
```

__Find the single most important word__

```python
import numpy as np

scores = X.toarray()[0]
features = vectorizer.get_feature_names_out()

index = np.argmax(scores)

print("Most important word:", features[index])
print("Score:", scores[index])
```
