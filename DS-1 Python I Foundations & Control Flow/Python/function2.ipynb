#### Positional and keyword arguments
- Python functions can contain two types of arguments: positional arguments and keyword arguments. Positional arguments must be included in the correct order. Keyword arguments are included with a keyword and equals sign.
- An argument is a variable, value or object passed to a function or method as input. Positional arguments are arguments that need to be included in the proper position or order.

- The first positional argument always needs to be listed first when the function is called. The second positional argument needs to be listed second and the third positional argument listed third, etc.

- A keyword argument is an argument passed to a function or method which is preceded by a keyword and an equals sign. The general form is:

`function(keyword=value)`




```python
from math import comb

def combination(num1, num2):
    result = comb(num1, num2)
    message = f"{num1} combination {num2} is: {result} "
    return print(message)

combination(8, 3)
```

#### Painting the Wall

We need to paint a wall and we do not know how many cans of paint we need to buy. The instructions on the paint can says that 1 can of paint can cover 4 square meters of wall. So we need to define a function which takes as parameter  height and width of wall and  calculates the area of the wall and based on the area we can calculate number of cans of paint that we need.

Area of wall = wall height  *  wall width

Number of cans of paint that is needed =  Area of wall ÷ coverage per can.

**Example**

Height = 2, Width = 5, Coverage = 4

Area of wall = 2  *  5  = 10

Number of cans of paint that is needed =  10 ÷ 4 = 2.5

But because you can't buy 2.5 of a can of paint, the result should be rounded up to 3 cans.

Hint: To round up number ceil function from math module can be used. Math.ceil()

```python
# @title
import math

def calculate_can_number(height, width, coverage):
    area = height * width
    number_of_cans = area / coverage
    return math.ceil(number_of_cans)


calculate_can_number(2, 5, 4)
```

#### Exercise
#### Calculator

Implement code block which asks two numbers and operation (+,-,*,/) that we want to perform on these numbers and returns the result.

**For example:**

- What is the first number? 2
- What is the second number? 4
- Pick operation from this list (+,-,*,/) *


```python
# @title
def add(num1, num2):
    return num1 + num2

def subtract(num1, num2):
    return num1 - num2

def multiply(num1, num2):
    return num1 * num2

def divide(num1, num2):
    return num1 / num2

num1 = int(input("What is the first number? "))
num2 = int(input("What is the second number? "))
operator = input('Pick operation from this list (+,-,*,/) ')

def calculate(num1, num2, operator):
    if operator == '+':
        result = add(num1, num2)
    elif operator == '-':
        result = subtract(num1, num2)
    elif operator == '/':
        result = divide(num1, num2)
    else:
        result = multiply(num1, num2)
    return result


answer = calculate(num1, num2, operator)
print(f"{num1} {operator} {num2} = {answer}")
```

#### Leap Year Rules: How to calculate leap year

- In the Gregorian calendar, there are three criteria that must be taken into account to identify a leap year.

1. The year must be evenly divisible by 4.
2. The year must also be evenly divisible by 400.
3. The year is not evenly divisible by 100.

```python
def leap(year):
    if (year % 400 == 0) or (year % 100 != 0) and (year % 4 == 0):
        return 'Leap Year'
    else:
        return 'Not leap year'

years = [1900, 2000, 2016, 1987, 1952]
for year in years:
    print(leap(year))
```

#### Gross pay
- Create a function called compute_pay which takes two parameters (hours and rate).
- Additionally, we need to create another function which checks if the type of "input" is a float or not.
- If the values for hour or rate is not float we will return error.

```python
def compute_pay(hours, rate):
    if hours < 40:
        pay = hours * rate
    else:
        overtime = hours - 40
        pay = (40 * rate) + (overtime * rate * 1.5)
    return pay

def is_float(value):
    try:
        val = float(value)
        return val
    except ValueError:
        print('Error, please enter a numeric value')
        quit()


hour = input('Enter hours: ')
hour = is_float(hour)
rate = input('Enter rate: ')
rate = is_float(rate)

amount = compute_pay(hour, rate)
print(f"Your gross pay is ${amount:.2f}")
```

#### Cold, Warm and Hot
Define a function which takes a temperature as a parameter:
- returns Hot if the temperature is greater than 28
- returns Warm if the temperature is between 18 and 28, including both.
- returns Cold if the temperature is less than 18

```python
def check_temp(temp):
    if temp > 28:
        return 'Hot'
    elif temp >= 18 and temp <= 28:
        return 'Warm'
    else:
        return 'Cold'

temp = [30, 16, 22, 29, 50, 14, 18]
for t in temp:
    print(f"It is {check_temp(t)} at {t}")
```
