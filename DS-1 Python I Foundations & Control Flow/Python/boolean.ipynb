### Boolean
- Boolean data type is a type of data that has only two possible values, `True` of `False`. In Python, any expression can be evaluated to either True or False. When two values are compared, the expression is evaluated and python returns the Boolean answer.

```python
print(True == True)
```

```python
print(7 == -7)
```

```python
print(3 != 5)
```

```python
print('mat' != 'mat')
```

```python
print(8 > 6)
```

```python
print(True and True)
```

```python
num1 = 5
num3 = 4
print(num1 is not num3)
```

```python
name = 'Mathew'
name = name.lower()
print('m' in name)
```
