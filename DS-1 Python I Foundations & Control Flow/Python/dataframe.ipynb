```python
planet = pd.DataFrame({
    'Planet':['Mercury', 'Venus', 'Earth', 'Mars', 'Jupiter', 'Saturn', 'Uranus', 'Neptune'],
    'radius_km': [2490, 6052, 6371, 33390, 69911, 58232, 25362, 24622],
    'Moons': [0, 0, 1, 2, 80, 83, 27, 14]
})

planet
```

```python
clothes = pd.DataFrame({'type': ['pants', 'shirt', 'shirt', 'pants', 'shirt', 'pants'],
                       'color': ['red', 'blue', 'green', 'blue', 'green', 'red'],
                       'price_usd': [20, 35, 50, 40, 100, 75],
                       'mass_g': [125, 440, 680, 200, 395, 485]})


clothes
```

```python
clothes.groupby(['type', 'color']).size()
```

```python
clothes.groupby('color').agg({'price_usd': ['mean', 'max'],
                             'mass_g': ['mean', 'max']})
```

```python
grouped = clothes.groupby(['color', 'type']).agg(['mean', 'min'])
grouped
```

```python
data = {'EmployeeID':[100, 101, 102, 103, 104, 105, 106, 107, 108, 109],
    'Profession':['Data Analyst', 'Data Scientist', 'Software Engineer',
              'Data Architect','Data Analyst', 'Business Analyst', 'Data Analyst',
              'Data Scientist', 'Software Engineer', 'Data Analyst'],
    'Firstname':['Mary', 'Sarah', 'John', 'Abel', 'Abaraham',
             'Mbappe', 'Elon', 'Chloe', 'Stone', 'Green'],
   'Lastname':['Smith', 'Jones', 'wick', 'Light', 'Blessing',
            'Saul', 'Orbita', 'Mat', 'Kelvin', 'Dino'],
    'Age':[24, 27, 30, 26, 25, 28, 29, 32, 35, 28],
    'Salary':['$50,000', '$75,000', '$80,000', '$65,000', '$55,000', '$60,000',
          '$70,000', '$85,000', '$90,000', '$73,000']}

employee = pd.DataFrame(data)
```

```python
# Create a column call Name
employee['Name'] = employee['Firstname'] + ' ' + employee['Lastname']
employee.head(3)
```

```python
employee = employee[['EmployeeID', 'Profession', 'Firstname', 'Lastname', 'Name', 'Age', 'Salary']]
employee.head(2)
```

```python
employee['Salary'] = (employee['Salary'].str.replace('$', '', regex = False)
                    .str.replace(',', '')
                    .astype('int'))

```

```python
employee.head(2)
```

```python
employee.dtypes
```

```python
print(employee.Salary.mean())
```

```python
print(f"Average salary of the employee is ${employee.Salary.mean():,.0f}")
```

```python
print('Maximum Salary:', employee['Salary'].max())
print('Minimum salary:', employee['Salary'].min())
print('Number of employee:', employee['Salary'].count())
```

```python
avg_salary_by_profession = employee.groupby('Profession')['Salary'].mean()
avg_salary_by_profession
```

```python
employee_agg = employee.groupby('Profession')['Salary'].agg(['count', 'mean', 'max', 'min'])
employee_agg
```
