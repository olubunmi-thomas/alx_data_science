### `Syntax and Semantics`

- `Syntax:` the rules that describe valid combinations of Python symbols

- `Semantics:` the meaning of a combination of Python symbols is the meaning of an instruction — what a particular combination of symbols does when you execute

### `Keyword Argument`
The print() function has two keyword arguments that can be very useful. They are `end` and `sep`

Basically, a keyword has three elements:
1. a keyword identifying the argument
2. an equal sign (=) and 
3. value assigned to that argument

Any keyword arguments must be placed after the last positional argument

### `The keyword end`
It specifies what to print at the end of the print statement

```python
print("Python is fun!", "I love Python programming language", end="")
```

```python
print("Python is fun!", end=" ")
print("I love Python programming language")
```

```python
for k in range(0, 3):
    print("python", end = " ")
```

### `The Keyword sep`
- It specifies the separator between the outputted arguments 

```python
first_name = "Sylvanus"
last_name = "Mark"
print(first_name.lower(), last_name.lower(), sep="_")
```

```python
print("Employee_records", "csv", sep=".")
```

```python
print('|' * 10)
```

### `Variable`
One of the most powerful features of a programming language is the ability to manipulate
variables. A variable is a name that refers to a value

**Rules of variables**
1. Don't use keywords or functions
2. spaces are not allowed
3. It can start with a letter or an underscore
4. It cannot start with a number
5. Must be made of letters, numbers or underscores

```python
print(help("keywords"))
```

```python
price = 10
print(price)
```

```python
price = 234
print(f"My mobile phone cost ${price}")
```

```python
x, y, z = 3, 6, 9
print(x)
print(y)
print(z)
```
