- A set is a special list that can only accept unique elements.
- Set can not be indexed
- Set doesn't have order
- set is mutable

### `Set creation`
As set can be created using curly bracket `{}`

```python
animal = {"aligator", "badger", "dog", "eagle", "fox", "goat"}
print(animal)
```

```python
print(type(animal))
```

#### Set does not allow duplicate

```python
cars = {"Mazda", "Mercedes", "Mazda", "Bughatti", "Ford", "Ford"}
print(cars)
```

`Note:`
- The values True and 1 are considered the same value in sets, and are treated as duplicates:

```python
my_tuple = {'Ford', 'Mazda', True, 3.9, 4, 1}
print(my_tuple)
```

#### Get the Length of a Set

To determine how many items a set has, use the `len()` function.

```python
cars = {"Mazda", "Mercedes", "Bughatti", "Ford"}
print(len(cars))
```

**Adding element to a set**
- The function `add` can be used to add elememt to a set

```python
cars = {"Mazda", "Mercedes", "Bughatti", "Ford"}
cars.add("Honda")
print(cars)
```

#### Add Any Iterable

The object in the `update()` method does not have to be a set, it can be any iterable object (tuples, lists, dictionaries etc.).

```python
cars = {"Mazda", "Mercedes", "Bughatti", "Ford"}
my_list = ['Lambogini', 'Honda']
cars.update(my_list)
print(cars)
```

#### Remove Item

To remove an item in a set, use the `remove()`, or the `discard()` method.

```python
cars = {"Mazda", "Mercedes", "Bughatti", "Ford"}
cars.remove('Bughatti')
print(cars)
```

```python
cars = {"Mazda", "Mercedes", "Bughatti", "Ford"}
cars.discard('Bughatti')
print(cars)
```

**Note:**
-  If the item to remove does not exist, `discard()` will NOT raise an error.

```python
cars = {"Mazda", "Mercedes", "Bughatti", "Ford"}
cars.discard('Honda')
print(cars)
```

#### Clear()
- The `clear()` method empties the set:

```python
cars = {"Mazda", "Mercedes", "Bughatti", "Ford"}
cars.clear()
print(cars)
```

#### del keyword
- The del keyword will delete the set completely:

```python
cars = {"Mazda", "Mercedes", "Bughatti", "Ford"}
del cars
```

#### The `set()` Constructor

It is also possible to use the `set()` constructor to make a set.

```python
cars = ("Mazda", "Mercedes", "Bughatti", "Ford")
print(set(cars))
```

```python
numbers = [2, 2, 2, 2, 3, 5, 9, 8, 5, 5]
print(set(numbers))
```

#### Access Items

You cannot access items in a set by referring to an index or a key.

But you can loop through the set items using a for loop, or ask if a specified value is present in a set, by using the in keyword.

```python
cars = {"Mazda", "Mercedes", "Bughatti", "Ford"}

if 'Mazda' in cars:
    print('Yes, Mazda is in cars')
```

```python
cars = {"Mazda", "Mercedes", "Bughatti", "Ford"}

for car in cars:
    print(car.lower())
```

#### Join Two Sets

There are several ways to join two or more sets in Python.

You can use the `union()` method that returns a new set containing all items from both sets, or the `update()` method that inserts all the items from one set into another:

```python
cars = {"Mazda", "Mercedes", "Bughatti", "Ford"}
Cars = {'Honda', 'Mazda', 'Ford'}
```

```python
cars.update(Cars)
print(cars)
```

```python
cars.union(Cars)
print(cars)
```

#### Pipe `|` can be used to perform the same task s `union()`

```python
cars = {"Mazda", "Mercedes", "Bughatti", "Ford"}
Cars = {'Honda', 'Mazda', 'Ford'}
cars |= Cars
print(cars)
```

## Keep ONLY the Duplicates

The `intersection_update()` method will keep only the items that are present in both sets.

```python
cars = {"Mazda", "Mercedes", "Bughatti", "Ford"}
Cars = {'Honda', 'Mazda', 'Ford'}
cars.intersection_update(Cars)
print(cars)
```

#### intersection()
- The `intersection()` method will return a new set, that only contains the items that are present in both sets.

```python
cars = {"Mazda", "Mercedes", "Bughatti", "Ford"}
Cars = {'Honda', 'Mazda', 'Ford'}
common = cars.intersection(Cars)
print(common)
```

`&` can be used to obtain the `intersection()` of two sets.

```python
cars = {"Mazda", "Mercedes", "Bughatti", "Ford"}
Cars = {'Honda', 'Mazda', 'Ford'}
common = cars & Cars
print(common)
```

#### Keep All, But NOT the Duplicates

- The `symmetric_difference_update()` method will keep only the elements that are NOT present in both sets.

```python
cars = {"Mazda", "Mercedes", "Bughatti", "Ford"}
Cars = {'Honda', 'Mazda', 'Ford'}
cars.symmetric_difference_update(Cars)
print(cars)
```

- The `symmetric_difference()` method will return a new set, that contains only the elements that are NOT present in both sets.

```python
cars = {"Mazda", "Mercedes", "Bughatti", "Ford"}
Cars = {'Honda', 'Mazda', 'Ford'}
new_cars = cars.symmetric_difference(Cars)
print(new_cars)
```

Your task for this quiz is to find the number of unique words in the text. In the code editor below, complete these three steps to get your answer.

```python
verse = "if you can keep your head when all about you are losing theirs and blaming it on you if you can trust yourself when all men doubt you but make allowance for their doubting too if you can wait and not be tired by waiting or being lied about don’t deal in lies or being hated don’t give way to hating and yet don’t look too good  nor talk too wise"
print(verse, '\n')

# split verse into list of words
verse_list = verse.split()
print(verse_list, '\n')

# convert list to a data structure that stores unique elements
verse_set = set(verse_list)
print(verse_set, '\n')

# find the number of unique words
num_unique = len(verse_set)
print(num_unique, '\n')
```

```python
def count_words(words):
    result = {}
    for word in words:
        if word not in result:
            result[word] = 0
        result[word] += 1
    return result


verse = "if you can keep your head when all about you are losing theirs and blaming it on you if you can trust yourself when all men doubt you but make allowance for their doubting too if you can wait and not be tired by waiting or being lied about don’t deal in lies or being hated don’t give way to hating and yet don’t look too good  nor talk too wise"
values = verse.split()
verse_dict = count_words(values)
print(verse_dict)
```
