```python
def game():
    player_choice = input('Enter a choice [rock, card, die]: ')
    options = ['rock', 'card', 'die']
    computer_choice = choice(options)
    choices = {'player':player_choice,
               'computer':computer_choice}
    print(choices)
    if player_choice == computer_choice:
        print("You guess correctly")
    else:
        print("You missed it. Try again!")

game()
```

### `Task`

|Points | Prize|
|-------|---- |
|1 - 50 | wooden rabbit|
|51 - 150 | no prize|
|151 - 180 | wafer-thin mint|

All of the lower and upper bounds here are inclusive, and points can only take on positive integer values up to 200.

You decide you want to play a game where you are hiding a number from someone. Store this number in a variable called 'answer'. Another user provides a number called 'guess'. By comparing 'guess' to 'answer', you inform the user if their guess is too high or too low. Fill in the conditionals below to inform the user about how their guess compares to the answer.

```python
answer = int(input('Enter a number: '))
guess = int(input('Enter a guess: '))

if guess < answer:
    result = "Oops!  Your guess was too low."
elif guess > answer:
    result = "Oops!  Your guess was too high."
else:
    result = "Nice!  Your guess matched the answer!"

print(result)
```

#### Tax Purchase

Depending on where an individual is from we need to tax them appropriately. The states of CA, MN, and NY have taxes of 7.5%, 9.5%, and 8.9% respectively. Use this information to take the amount of a purchase and the corresponding state to assure that they are taxed by the right amount.

```python
state = "CA"  #Either "CA", "MN", or "NY"
purchase_amount = 1200.34

if state == 'CA':
    tax = .075
    total_cost = purchase_amount*(1+tax)
    result = f"Since you're from {state}, your total cost is ${total_cost:,.2f}."

elif state == 'MN':
    tax = .095
    total_cost = purchase_amount*(1+tax)
    result = f"Since you're from {state}, your total cost is ${total_cost:,.2f}."

elif state == 'NY':
    tax = .089
    total_cost = purchase_amount*(1+tax)
    result = f"Since you're from {state}, your total cost is ${total_cost:,.2f}."

print(result)
```

#### Leap Year Rules: How to calculate leap year

- In the Gregorian calendar, there are three criteria that must be taken into account to identify a leap year.

1. The year must be evenly divisible by 4.
2. The year must also be evenly divisible by 400.
3. The year is not evenly divisible by 100.




```python
year = int(input('Enter a year '))

if (year % 400 == 0) or (year % 100 != 0) and (year % 4 == 0):
    print('Leap Year')
else:
    print('Common year')
```

```python
def isLeap(year):
    if (year % 400 == 0) or (year % 100 != 0) and (year % 4 == 0):
        return 'Leap Year'
    else:
        return 'Common year'


year = [1900, 2000, 2016, 1987]
for y in year:
    print(isLeap(y))
```
