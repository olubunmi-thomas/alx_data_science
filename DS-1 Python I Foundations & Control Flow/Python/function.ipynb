list of parameters representing the data that the function needs to perform its task.

In python, there are two types of function:
- Function can:

#### Count of odd and even numbers

```python
def count(num1, num2):
    odd_numbers = 0
    even_numbers = 0
    
    for number in range(num1, num2):
        if number % 2 != 0:
            odd_numbers += 1
        else:
            even_numbers += 1
    return f"There are {odd_numbers} odd numbers \nThere are {even_numbers} even numbers"
    
    
print(count(1, 31))
```

#### Function to create list of odd or even numbers

```python
def odd_or_even(a, b):
    odd_numbers = []
    even_numbers = []
    for number in range(a, b):
        if number % 2 != 0:
            odd_numbers.append(number)
        else:
            even_numbers.append(number)
    return f'Odd numbers are: {odd_numbers} \nEven numbers are: {even_numbers}'
    
        
print(odd_or_even(1, 23))           
```

```python
def vowel_eater(word):
    vowel:str
    word_without_vowel = ""
    vowels = ['A', 'E', 'I', 'O', 'U']
    word = word.upper()

    for letter in word:
        if letter in vowels:
            continue
        word_without_vowel += letter
    return word_without_vowel


print(vowel_eater('gregory'))
print(vowel_eater('abstemious'))
print(vowel_eater('starvation'))
print(vowel_eater('glorification'))
```

```python
from statistics import mean

def sum_score_above_average(p_student_scores):
    average = mean(p_student_scores)
    total = 0
    for score in p_student_scores:
        if score > average:
            total += score
    print(total)
    

student_scores = [80, 60, 50, 65, 75, 55]
sum_score_above_average(student_scores)
```

### `Palindrome`
A palindrome is a string that can be equally read from left to right or right to left, omitting blank spaces, and ignoring capitalization. Examples of palindromes are words like kayak, madam and radar, and phrases like "Never Odd or Even".

```python
def is_palindrome(input_string):
    new_string = input_string.lower().replace(" ","")
    if new_string == new_string[::-1]:
        return True
    else:
        return False

print(is_palindrome("Never Odd or Even"))
print(is_palindrome("abc"))
print(is_palindrome("kayak"))
print(is_palindrome("madam"))
```

### `Exercise`
Write a function that accepts an input list and returns a new list which contains only the unique elements and the other of the list must be preserved as the original list.

```python
def unique(lst):
    new_lst = []
    for val in lst:
        if val not in new_lst:
            new_lst.append(val)
    return new_lst

print(unique([1, 1, 2, 4, 4, 6, 8, 9, 3, 5]))
```

```python
def sort_list(lst):
    lst.sort()
    return print(lst)

sort_list(['lion', 'jackal', 'leopard', 'horse', 'donkey', 'crab', 'badger', 'alligator'])
sort_list([18,25,3,41,5])
```

```python
def sort_descend(lst):
    lst.sort(reverse = True)
    return print(lst)

sort_descend(['alligator', 'badger', 'crab', 'donkey', 'horse', 'jackal', 'leopard', 'lion'])
sort_descend([18,25,3,41,5])
```

### `A function to perform the previous two exercises`

```python
def sort_list(lst, reverse = False):
    if reverse == True:
        lst.sort(reverse = True)
    else:
        lst.sort()
    return lst

sort = sort_list(['lion', 'jackal', 'leopard', 'horse', 'donkey', 'crab', 'badger', 'alligator'])
print(sort)
print(sort_list([18,25,3,41,5], reverse = True))
```

### `args`
- Highly useful when you don't know the number of argument your function will take.

```python
def multiply(*args):
    product = 1
    for number in args:
        product *= number
    return product


print(multiply(2, 3, 4, 5))
```

```python
def add(*args):
    total = 1
    for number in args:
        total += number
    return total


print(add(6, 20, 23, 45, 67))
```

#### Task
Given n, take the sum of the digits of n. If that value has more than one digit, continue reducing in this way until a single-digit number is produced. The input will be a non-negative integer.
Examples

- 45  -->  4 + 5 = 9
- 942  -->  9 + 4 + 2 = 15  -->  1 + 5 = 6
- 376 -->  3 + 7 + 6 = 16 --> 1 + 6 = 7

```python
def digital_root(n):
    x = str(n)
    a = n
    while len(x) > 1:
        a = 0
        for i in range(len(x)):
            a += int(x[i])
        x = str(a)
    return a


print(digital_root(45))
print(digital_root(942))
print(digital_root(376))
```

#### Task
Factorial

```python
def factorial(number):
    if number == 0 or number == 1:
        return 1
    else:
        product = 1
        # write your for loop here
        for n in range(2, number + 1):
            product *= n
        return f"{number}! is {product}"

print(factorial(5))
```
