#### Class work:
- Create a function call 'remove_ashtag'
- with a parameter sentence, then remove all the ashtag in the sentence.
-  e.g. '#Class of 97 #I #love you all#'

```python
# @title
def remove_ashtag(sentence):
    sentence = sentence.split('#')
    sentence = ''.join(sentence)
    return sentence.strip()


text = '#Class of 97 #I #love you all#'
print(remove_ashtag(text))
```

```python
def split_by_ashtag(sentence):
    sentence = sentence.split('#')
    return sentence


text = '#Class of 97 #I #love you all#'
print(split_by_ashtag(text))
```

```python
def sum_all(number1, number2):
    total = 0
    for number in range(number1, number2):
        total += number
    return total

```

#### Task
Write a function  that will do the following:
- If John saves `100 dollars`, he will get additional `10 dollars`
- otherwise, tell John to `save more!`

```python
def add_10(saving):
    if saving >= 100:
        return saving + 10
    else:
        return 'Save more!'


# Call the function
savings = [120, 96, 74, 102, 88]
for saving in savings:
    print(add_10(saving))
```

```python
def add_10(saving):
    if saving >= 100:
        return saving + 10
    else:
        return 'Save more!'


# Call the function
result = []
savings = [120, 96, 74, 102, 88]
for saving in savings:
    result.append(add_10(saving))
print(result)
```

#### List comprehension

```python
def add_10(amount):
    if amount >= 100:
        return amount + 10
    else:
        return 'Save more!'


savings = [120, 96, 74, 102, 88]
result = [add_10(saving) for saving in savings]
print(result)
```

#### Task
- Write a function called `is_even` that checks whether a number is even. Return `True` if the number is even and `False` otherwise.

- Test with this list: [2, 13, 42, 37, 97, 104]

```python
def is_even(number):
    if number % 2 == 0:
        return True
    return False


numbers =  [2, 13, 42, 37, 97, 104]
for n in numbers:
    print(is_even(n))
```
