__Task 7__
> There are times when your answer is a double data type and you want to show the same in scientific format. Here, we have to get a scientific format answer out of a double data type. You have to input  a and need to output its scientific format upto 4 precision.

```python
def scientific_notation(a:float):
    return f"{a:.4e}"

print(scientific_notation(3.4))
```

__Task 8__

> There are times when your answer is a floating point that contains undesired amount of digits after decimal.
> Here, we'll learn how to get a precise answer out of a floating number. You are given two floating numbers a and b.

> - You need to print the value of a/b rounded to exactly 3 decimal places.
> - You may use library precision methods.
> - You need to print the result (return type is void) without any spaces around it.

```python
def floating(a, b):
    result = a / b
    return f"{result:.3f}"

print(floating(2.45, 3.97))
```

__Task 9__

> Given an integer a, convert the decimal number a into its hexadecimal representation and print the answer.

```python
def hexadecimal(a):
    return f"{a:.02x}" # Pad with zeros to a width of 2 digits

# code here
print(f"{26:02x}")
print(f"{20:02x}")
```
