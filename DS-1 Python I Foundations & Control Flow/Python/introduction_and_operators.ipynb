#### Variable
A variable is a named container which stores values in a reserved location in the computer’s memory.

#### Rules for Python Variables
1. Python keywords can not be used as a variable name.
2. A variable name must start with a letter or an underscore character.
3. A variable cannot start with a number.
4. A variable name can only contain alpha-numeric characters and underscores.
5. Special characters such as '@', '$', '%', etc are not allowed.
6. Variable names are case sensisitive. For instance, Name, name, NaMe, naMe are four different varaible names.

The most commonly used methods of constructing a multi-word variable name are:
1. **Camel Case:** Second and subsequent words are capitalized, to make word boundaries easier to see. For example, averageCost
2. **Pascal Case:** Indentical to `camel case` but the first word is also capitalized. For example, AverageCost
3. **Snake Case:** Words are separated by underscores. For example, average_cost.

According to PEP 8 naming conventions;
- **Snake Case** should be used for functions and variable names.
- **Pascal Case** should be used for class names.

#### Logical operators

Python also has three logical operators that can be combined with comparators to create more complex statements.

These operators are:

2. `or`
