## `Python Data Types`
Python Data Types are used to define the type of a variable. It defines what type of data we are going to store in a variable.

- Integer- Whole numbers
- Float - Number with decimal fraction
- complex- Imaginary numbers
- Boolean - Are True or False
- String - Sequence of characters that are under single qoutes or double qoutes


### `Python numbers`
There are three data type in python
- integer
- float
- complex

```python
a = 10
b = 1.43
c = 2 + 4j
```

 **The `type()` function can be used to verify the type of any object in Python.**

```python
print(type(a))
print(type(b))
print(type(c))
```

### `Integer`
- This is a whole number, can be positive or negative, without decimals, of unlimited length.

```python
a = 8
b = 23145723
c = -3421
```

```python
print(type(a))
print(type(b))
print(type(c))
```

### `Float`
- Float, or `floating point number` is a number, positive or negative, containing one or more decimals.

```python
a = 9.0
b = 12.86
c = -11234.50
```

```python
print(type(a))
print(type(b))
print(type(c))
```

- Float can also be expressed in scientific notation with an `e`  or `E`to indicate the power of 10.
- For example `4e3` = 4 x $ 10^3 $

```python
a = 2e6
b = -18e5
c = 1e-3
```

```python
print(type(a))
print(type(b))
print(type(c))
```

### `String operations`

- `len(string)` - Returns the length of the string
- `for character` in string - Iterates over each character in the string
- `if substring in string` - Checks whether the substring is part of the string
- `string[i]` - Accesses the character at index i of the string, starting at zero
- `string[i:j]` - Accesses the substring starting at index i, ending at index j minus 1. If i is omitted, its value defaults to 0. If j is omitted, the value will default to len(string).

### `len()`

```python
state = 'colorado'
print(len(state))
```

#### `String Concatenation:`

`+` operator

In Python, we can join (concatenate) two or more strings using the + operator.

```python
text ="I'll be in Colorado"
day = 'tomorrow'
print(text + " " + day)
```

```python
last_name = "smith"
first_name = "stone"
domain = "@prime.info"
email = first_name + last_name + domain
print(email)
```

#### `Using f-string for concatenation`

```python
last_name = "smith"
first_name = "stone"
domain = "@prime.info"
email = f"{first_name}{last_name}{domain}"
print(email)
```

#### ` Using sep to concatenate`


```python
last_name = "smith"
first_name = "stone"
domain = "@prime.info"
print(first_name,last_name,domain,sep='')
```

#### `Iterate Through a Python String`

We can iterate through a string using a for loop

```python
state = 'Colorado'
for letter in state:
    print(letter)
```

```python
state = 'Colorado'
for letter in state:
    print(letter, end='')
```

```python
state = 'Colorado'
for i in range(3):
    print(state)
```

```python
state = 'Colorado'
for i in range(3):
    print(state, i+1)
```

#### `break and continue`

```python
state = 'Colorado'
for letter in state:
    if letter == 'r':
        break
    print(letter)
```

```python
state = 'Colorado'
for letter in state:
    if letter == 'r':
        continue
    print(letter)
```


#### `String Method`

|Methods | Description|
|--------| -------- |
|upper( )| converts the string to uppercase|
|lower( )|converts the string to lowercase|
|partition( )|returns a tuple|
|find( )|returns the index of first occurrence of substring|
|strip( )|removes trailing characters|
|string.lstrip()|Returns a copy of the string with the left-side whitespace removed|
|string.rstrip() |Returns a copy of the string with the right-side whitespace removed|
|split( )|splits string from left|
|startswith( )|checks if string starts with the specified string|
|endswith( )|checks if string starts with the specified string|
|isnumeric( )|checks numeric characters|
|index( )|returns index of substring|
|string.count(substring)|Returns the number of times substring is present in the string|
|string.isalpha()|Returns True if there are only alphabetic characters in the string. If not, returns False.|
|string.split(delimiter)| Returns a list of substrings that were separated by whitespace or a delimiter|
|string.replace(old, new)|Returns a new string where all occurrences of old have been replaced by new.|
|delimiter.join(list of strings)| Returns a new string with all the strings joined by the delimiter|
|swapcase()| swaps the letters' cases (lower to upper and vice versa)|
|title()|makes the first letter in each word upper-case|

#### `join()`

```python
numlist = ['1', '3', '5', '7', '9']
sep = '  '

print('List before Join():', numlist)
print('List after Join():' , sep.join(numlist))
```

```python
names = ['Smith', 'Stone']
sep = '_'
user_name = sep.join(names)
print(user_name.lower())
```

```python
words = ['Hello','can','we', 'play', 'today', '!']
sep = ' '
sentence = sep.join(words)

print(sentence)
```

```python
# .join with string
alpha = 'ABC'
numeric = '123'

print(alpha.join(numeric))
print(numeric.join(alpha))
```
