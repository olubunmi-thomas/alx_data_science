## `Using python as calculator`

## `Function to multiply two numbers`

```python
def multiply(a, b):
    """This function find the product of two numbers"""
    result = a * b
    return print('The product of {} and {} is {}'.format(a, b, result))
```

```python
# Call the fuction
multiply(4, 16)
```

```python
def product(num1, num2):
    result = num1 * num2
    return print(f'The product of {num1} and {num2} is {result}')
```

```python
# Call the fuction
product(4, 16)
```

## `Function to add two numbers`

```python
def add_two(num1, num2):
    """This function evaluate the sum of two numbers"""
    Sum = num1 + num2
    return print(f"The sum of {num1} and {num2} is {Sum}")
```

```python
# Call the function
add_two(5, 8)
```

## `Function to find difference between two numbers`

```python
def subtract(num1, num2):
    """This function evaluate difference between two numbers"""
    difference = num1 - num2
    return print(f"{num1} minus {num2} is {difference}")
```

```python
# Call the function
subtract(5, 8)
```

## Function to find positive difference between two numbers

```python
def positive_difference(num1, num2):
    """This function take two numbers and obtain the positive difference"""
    if num1 > num2:
        result = num1 - num2
    else:
        result = num2 - num1
    return print(f"The positive difference between {num1} and {num2} is {result}")
```

```python
positive_difference(8, 17)
```

```python
positive_difference(98, 86)
```

## Function to find the average of numbers

```python
import numpy as np

def average(my_list):
    total = sum(np.array(my_list))
    result = total / len(my_list)
    return print(result)
```

```python
my_list = [2, 3, 5, 7, 11]
average(my_list)
```

```python
my_list = range(0, 25, 3)
average(my_list)
```

## Function to extract string in a list

```python
def extract_string(lst):
    for animal in lst:
        if type(animal) == str:
            print(animal)
```

```python
mixed_list = [2.3, 'cat', 'cow', 4, 8.9, 'donkey', 'monkey', 6]
extract_string(mixed_list)
```

### `Task:`

- Write a function that accepts two positive integers which are length and width of a rectangle and retuns a list that contains the area and perimeter of that rectangle.


```python
def area_perimeter(length, width):
    area = length * width
    perimeter = 2 * (length + width)
    result = [area, perimeter]
    return print(result)

area_perimeter(4, 6)
```

### `Task:`
- Write a function to calculate the area of a circle

```python
from math import pi
```

```python
def area_of_circle(radius):
    area = pi * radius ** 2
    return print(f"The area of circle radius {radius}cm is {area:.2f}")

area_of_circle(7)
    
```

## `Function to convert strings in a list to string`

```python
def convert_to_lower(my_list):
    lst = [i.lower() for i in my_list]
    print(lst)
    
convert_to_lower(['Mark', 'PauL', 'JOhn', 'Lewis', 'Judah', 'PAULINA'])
```

- Change `i.lower()` to `i.capitalize()` in the above code and check what happen.

```python
def capitalize(my_list):
    lst = [i.capitalize() for i in my_list]
    print(lst)
        
capitalize(['Mark', 'PauL', 'JOhn', 'Lewis', 'Judah', 'PAULINA'])
```
