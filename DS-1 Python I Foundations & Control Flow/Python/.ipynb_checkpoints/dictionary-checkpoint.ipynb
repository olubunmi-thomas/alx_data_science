## `Dictionary`
A dictionary is a mutable, unordered data structure that contains mappings of keys to values. Because these keys are used to index values, they must be unique and immutable. For example, a string or tuple can be used as the key of a dictionary, but if you try to use a list as a key of a dictionary, you will get an error.

Because dictionaries are not ordered, they are not sortable. And you do not add items to a dictionary with .append.

### `Note:`
In Python, any immutable object (such as an integer, boolean, string, tuple) is hashable, meaning its value does not change during its lifetime. This allows Python to create a unique hash value to identify it, which can be used by dictionaries to track unique keys and sets to track unique values. This is why Python requires us to use immutable datatypes for the keys in a dictionary.

```python
animal_sound = {
    "alligator":["bellow","hiss"],
    "badger":"grow",
    "crab":['chirp', 'click', 'creak'],
    "eagle":"screech",
    "hamster":"squeak",
    "jackal":"gecker"
}
print(animal_sound["eagle"])
```

#### get() method

```python
print(animal_sound.get('eagle'))
```

### `keys() and values()`

```python
animal_sound = {
    "alligator":["bellow","hiss"],
    "badger":"grow",
    "crab":['chirp', 'click', 'creak'],
    "eagle":"screech",
    "hamster":"squeak",
    "jackal":"gecker"
}
print(list(animal_sound.keys()))
```

```python
print(list(animal_sound.values()))
```

```python
s = list(animal_sound.values())
print(s[2])
```

#### Dictionary from tuples

```python
animals = ['Aligator', 'Badger', 'Crab', 'Eagle', 'Jackal']
sounds = ['Bellow', 'Grow', 'Click', 'Screech', 'Gecker']
id_no = [1, 2, 3, 4, 5]


animal_tup = list(zip(animals, sounds, id_no))
print(animal_tup)
```

```python
animal_dict = {}
for animal, sound, id in animal_tup:
    if animal in animal_dict:
        animal_dict[animal].append((sound, id))
    else:
        animal_dict[animal] = [(sound, id)]


print(animal_dict)
```

#### Get items()
- The `items()` method will return each item in a dictionary, as tuples in a list.

```python
animal_sound = {
    "alligator":["bellow","hiss"],
    "badger":"grow",
    "crab":['chirp', 'click', 'creak'],
    "eagle":"screech",
    "hamster":"squeak",
    "jackal":"gecker"}

print(animal_sound.items())
```

### `Ditionary comprehension`

```python
a = {k:v for k, v in animal_sound.items()}
print(a)
```

### `List comprehension`

```python
print([k for k in animal_sound.keys()])
```

```python
print([v for v in animal_sound.values()])
```

### `Tuple as key`

```python
room_numbers = {
    ('Freddie', 'Jen'): 403,
    ('Ned', 'Keith'): 391,
    ('Kristin', 'Jazzmyne'): 411,
    ('Eugene', 'Zach'): 395
}
print([key for key in room_numbers.keys()])
```

### `Exercise`
- Create an empty dictionary
- Iterate through each element in the list. If an element is already included in the dictionary, add 1 to its value. If not, add the element to the dictionary and set its value to 1.

```python
title =  ['great', 'expectations','the', 'adventures', 'of', 'sherlock','holmes','the',
            'great','gasby','hamlet','adventures','of','huckleberry','fin']

d = {}
for word in title:
    d[word] = d.get(word, 0) + 1

print(d)
```

### `Exercise`
The email_list function receives a dictionary, which contains domain names as keys, and a list of users as values. Fill in the blanks to generate a list that contains complete email addresses (e.g. diana.prince@gmail.com).

```python
def email_list(domains):
    emails = []
    for domain, users in domains.items():
        for user in users:
            emails.append(user+"@"+domain)
    return emails

print(email_list({"gmail.com": ["clark.kent", "diana.prince", "peter.parker"],
                  "yahoo.com": ["barbara.gordon", "jean.grey"],"hotmail.com": ["bruce.wayne"]}))
```
