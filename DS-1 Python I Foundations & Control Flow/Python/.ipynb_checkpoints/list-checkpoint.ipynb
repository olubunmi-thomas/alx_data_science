### `List`

List is like a container that holds a number of items. Each element or value in a list is called an item. All the items in a list are assigned to a single variable

- In python, list can be defined by using square brackets `[]`
- List are mutable and dynamics
- List are sortable, you can add an item to a list with .append
- list items are always indexed with numbers starting at 0
- Lists can contain any arbitrary objects.
- There is no conceptual limit to the size of a list

```python
animal = ['aligator', 'bear', 'cat', 'dog', 'eagle']
print(animal)
```

```python
print(type(animal))
```

```python
even = list(range(2, 31, 2))
print(even)
```

```python
multiple_of_5 = list(range(0, 51, 5))
print(multiple_of_5)
```

### `Looping through a list`

**for loop**

```python
vowels = ['a', 'e', 'i', 'o', 'u']
for vowel in vowels:
    print(vowel)
```

#### `To print all the items of the vowels list on the same line`

```python
vowels = ['a', 'e', 'i', 'o', 'u']
for vowel in vowels:
    print(vowel, end=' ')
```

```python
vowels = ['a', 'e', 'i', 'o', 'u']
for vowel in vowels:
    print(vowel, end='|')
```

### `break and continue`

```python
vowels = ['a', 'e', 'i', 'o', 'u']
for vowel in vowels:
    if vowel == 'o':
        break
    print(vowel)
```

```python
vowels = ['a', 'e', 'i', 'o', 'u']
for vowel in vowels:
    if vowel == 'o':
        continue
    print(vowel)
```

```python
numbers = range(2, 20)
for n in numbers:
    if n % 2 == 0:
        continue
    print(n)
```

### `Getting each item and its index`

```python
vowels = ['a', 'e', 'i', 'o', 'u']
for index, vowel in enumerate(vowels):
    print(index, vowel)
```

```python
vowels = ['a', 'e', 'i', 'o', 'u']
for v in vowels:
    capital = v.upper()
    print(v, capital,sep='|')
```

## List comprehension

```python
print([n for n in range(3, 15, 3)])
```

```python
cube = [c ** 3 for c in range(0, 30, 4)]
print(cube)
```

```python
vowels = ['a', 'e', 'i', 'o', 'u']
print([v.upper() for v in vowels])
```

```python
employee = ['primestute@Gmail.Com', 'masterTom@yaHoo.cOm', 'ChloEnik@mine.coM']
domain= [e.lower().split('@')[1] for e in employee]
print(domain)
```

```python
greet = 'Hello World!!'
chars = [c for c in greet]
print(chars)
```

```python
cars = ["Mazda", "Mercedes", "Lambogini", "Bughatti", "Toyota", "Ford", "GMC", "Royce", "Porche"]
print([len(l) for l in cars])
```

### `Exercise: `

- Filter Names by Scores
- Use a `list comprehension` to create a list of names `passed` that only include those that scored at `least 65`.

```python
scores = {"Martin Joe":90,
         "Morty Stone":64,
         "Summer Kelvin":76,
         "Jerry Smith":80,
         "Beth Cloe":59}
passed = [name for name, score in scores.items() if score >= 65]
print(passed)
```

### `Conditionals in List Comprehensions`
You can also add conditionals to list comprehensions (listcomps). After the iterable, you can use the `if` keyword to check a condition in each iteration.

```python
squares = [(i ** 2) for i in range(2, 19) if i % 2 != 0]
print(squares)
```

**If you would like to add else, you have to move the conditionals to the beginning of the list comprehension, right after the expression, otherwise, you will get error**

```python
cube = [(i ** 3) if i % 2 != 0 else i + 2 for i in range(2, 11)]
print(cube)
```

### `Exercise`
- Extract First Names
- Use a `list comprehension` to create a new list `first_names` containing just the first names in names in `lowercase.`

```python
names = ["Rick Sanchez", "Morty Smith", "Summer Smith", "Jerry Smith", "Beth Smith"]

first_names = [name.split(" ")[0].lower() for name in names]
print(first_names)
```

### `List Method`



### `list.Append()`
- Add a single element to the end of a list

```python
animal = ['aligator', 'bear', 'cat', 'dog', 'eagle', 'fox', 'giraffe']
animal.append('lemur')
print(animal)
```

### `list.insert()`
- Can be used to insert a single value to a specific position in a list

```python
animal = ['aligator', 'bear', 'cat', 'dog', 'eagle', 'fox', 'giraffe']
animal.insert(1,'badger')
print(animal)
```

```python
animal = ['aligator', 'bear', 'cat', 'dog', 'eagle', 'fox', 'giraffe']
animal.insert(-1,'goat')
print(animal)
```

### `list.pop()`
- The list.pop() method takes one parameter, an index i, and removes the value from the list at that index. The value that is removed is returned by the method. For example

```python
animal = ['aligator', 'bear', 'cat', 'dog', 'eagle', 'fox', 'giraffe']
animal.pop(3)
print(animal)
```

```python
animal = ['aligator', 'bear', 'cat', 'dog', 'eagle', 'fox', 'giraffe']
print(animal.pop(3))
```

## `Note:`
- The pop() removes the last item in the list if you do not pass a value into it.

```python
animal = ['aligator', 'bear', 'cat', 'dog', 'eagle', 'fox', 'giraffe']
print(animal.pop())
```

### `list.extend()`
The list.extend() method is used to add several new elements to the end of a list:

```python
animal = ['aligator', 'bear', 'cat', 'dog', 'eagle', 'fox', 'giraffe']
animal.extend(["lion", "monkey","moose"])
print(animal)
```

## `Note`
extend() takes a single parameter that must be an iterable type. The elements of the iterable are appended to the list in the same order that they appear in the argument passed to .extend().

## `list.sort()`
list.sort() sort the items of the list, in place.

```python
animal = ["monkey", "jackal", "tiger", "zebra", "aligator", "eagle"]
animal.sort()
print(animal)
```

```python
prime = [7, 11, 2, 5, 3, 13, 17, 19]
prime.sort(reverse=True)
print(prime)
```

```python
animal = ["monkey", "jackal", "tiger", "zebra", "aligator", "eagle"]
animal.sort(reverse = True)
print(animal)
```

### `list.reverse()`
The method reverses the items of the list in place.


```python
animal = ['aligator', 'bear', 'cat', 'dog', 'eagle', 'fox', 'giraffe']
animal.reverse()
print(animal)
```

## `list.count(a)`
This method returns the number of times a appear in the list

```python
sound = ["bellow", "mew", "hiss", "mew", "hiss", "hiss", "hum"]
print(sound.count("mew"))
```

## `list.index(a)`
This method returns the index in the list of the first item where value is a. It is an error if such item does not exist.

```python
animal = ["alligator", "badger", "crab", "eagle", "hamster", "jackal"]
crab_index = animal.index("crab")
print(crab_index)
```

## `list.remove(a)`
This method removes the first item from the list whose value is a.It is an error if such item does not exist.

```python
animal = ["alligator", "badger", "crab", "eagle", "hamster", "jackal"]
animal.remove("eagle")
print(animal)
```

## `list.clear()`
list.clear() removes all the items in the list.

```python
animal = ["alligator", "badger", "crab", "eagle", "hamster", "jackal"]
animal.clear()
print(animal)
```

### `del`

can be used to remove a range of elements

```python
animal = ["alligator", "badger", "crab", "eagle", "hamster", "jackal"]
del animal[0:3]
print(animal)
```

### Highest Score

- A List of scores of students are given, implement a program that calculates the highest score from the given list.

#### Example
- student_scores = [80, 60, 50, 65, 75, 55]

```python
student_scores = [80, 60, 50, 65, 75, 55]

student_scores.sort()
print(f'The highest score in the class is: {student_scores[-1]}')
```

#### Second method


```python
student_scores = [80, 60, 50, 65, 75, 55]

highest_score = 0
for score in student_scores:
    if score > highest_score:
        highest_score = score
print(f'The highest score in the class is: {highest_score}')
```
