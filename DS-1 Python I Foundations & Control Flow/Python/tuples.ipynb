## Tuples
Tuples are sequences just like lists. Tuples are defined by parentheses, `()`, and are immutable

```python
people = ('Paul', 'John', 'Chloe', 'Smith', 'Kelvin')
print(type(people))
```

#### Empty tuple
The empty tuple is written as two parentheses containing nothing `()`

```python
empty_tuple = ()
print(type(empty_tuple))
```

#### Tuple with single element


```python
single_tuple = ('Chole',)
print(type(single_tuple))
```

- Like string indices, tuple indices start at 0, and they can be sliced, concatenated, and so on.

#### indexing

```python
people = ('Paul', 'John', 'Chloe', 'Smith', 'Kelvin')
print(people[0])
```

```python
print(people[-1])
```

#### Slicing

```python
people = ('Paul', 'John', 'Chloe', 'Smith', 'Kelvin')
print(people[0:3])
```

```python
people = ('Paul', 'John', 'Chloe', 'Smith', 'Kelvin')
print(people[:4])
```

```python
people = ('Paul', 'John', 'Chloe', 'Smith', 'Kelvin')
print(people[3:])
```

```python
people = ('Paul', 'John', 'Chloe', 'Smith', 'Kelvin')
print(people[0:4:2])
```

#### Tuples allow Duplicate

```python
fruits = ("apple", "banana", "cherry", "apple", "cherry")
print(fruits)
```

#### Tuples allow different data types

```python
mixed_tuple = (1, 3.4, 'apple', 'mango', 6)
print(mixed_tuple)
```

#### Tuple length

The `len()` function can be used to determine the length of a tuple

```python
fruits = ("apple", "banana", "cherry", "apple", "cherry")
print(len(fruits))
```

#### Check if Item Exists

To determine if a specified item is present in a tuple use the `in` keyword:

```python
people = ('Paul', 'John', 'Chloe', 'Smith', 'Kelvin')

if 'Kelvin' in people:
    print('Yes')
else:
    print('No')
```

#### Unpacking a tuple

We can unpack a tuple by assigning variables to its elements

```python
prime = (2, 3, 5, 7, 11, 13, 17)
```

```python
first, *others, last = prime
```

```python
print(first)
print(last)
print(others)
```

```python
first, second, *others = prime
```

```python
print(first)
print(second)
print(others)
```

#### Loop through a tuple

```python
people = ('Paul', 'John', 'Chloe', 'Smith', 'Kelvin')

for p in people:
    print(p)
```

```python
people = ('Paul', 'John', 'Chloe', 'Smith', 'Kelvin')

for p in range(len(people)):
    print(people[p])
```

```python
people = ('Paul', 'John', 'Chloe', 'Smith', 'Kelvin')

for p in people:
    print(range(len(p)))
```

```python
people = ('Paul', 'John', 'Chloe', 'Smith', 'Kelvin')

for p in people:
    print(len(p))
```

#### Join Two Tuples

To join two or more tuples you can use the + operator:

```python
odd = (1, 3, 5, 7, 9)
even = (2, 4, 6, 8, 10)
numbers = odd + even
print(numbers)
```

#### Tuple Methods

Python has two built-in methods that you can use on tuples..

|Method | Description |
-------- | ------
|count()| Returns the number of times a specified value occurs in a tuple|
|index()| Searches the tuple for a specified value and returns the position of where it was found|

```python
fruits = ("apple", "banana", "cherry", "apple", "cherry")
print((fruits.count('apple')))
```

```python
print((fruits.index('banana')))
```
